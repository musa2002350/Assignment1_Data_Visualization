## Part B · Lie Factor, in code 

```python
import pandas as pd
import vizlib  # Ensure vizlib.py is in your working directory
```

```python

# 1. Load the Lie Factor dataset
df_lie = pd.read_csv("case07_lie_factor.csv")

# 2. Compute Lie Factor for each graphic
results = []
for idx, row in df_lie.iterrows():
    # Unpack the three values returned by the function tuple
    lf_val, data_eff, gfx_eff = vizlib.lie_factor(
        value_start=row['value_start'],
        value_end=row['value_end'],
        ink_start=row['ink_start'],
        ink_end=row['ink_end'],
        ink_dimension=row['ink_dimension']
    )
    
    results.append({
        'Graphic': row['chart_id'],
        'Data Effect': data_eff,
        'Graphic Effect': gfx_eff,
        'Lie Factor': lf_val,
        'Verdict': 'Honest' if 0.95 <= lf_val <= 1.05 else 'Distorted'
    })

df_results = pd.DataFrame(results)
display(df_results)
```

## Answers to Questions (1–5)

### 1 & 2. The Table and the Honest Chart
#### The Table: 
Produced via the code snippet above, showing data effect, graphic effect, Lie Factor, and verdict for charts A–E.  
#### The Honest Chart: 
Exactly one chart falls within Tufte’s acceptable range ($0.95$ to $1.05$). The other four distort reality by encoding values using 2D areas (like bubbles or squares) instead of 1D lengths, causing quadratic exaggeration.

### 3. Charts Needing ink_dimension=2
#### Identification: 
Charts that use scaling areas (circles/squares) instead of simple bar heights/lengths.
#### The Consequence: 
If you leave ink_dimension at 1, the formula assumes a linear 1D change, which fails to account for the square-cube law. Doubling the radius of a circle multiplies its area by $4$, drastically misrepresenting the true Lie Factor.

### 4. The Understated Chart (Lie Factor Below 1.0)

#### Why it's a distortion: 
Understating a trend minimizes real growth or decline, hiding important signals from decision-makers.

#### Who benefits: 
Usually organizations or publishers trying to downplay negative performance, regulatory scrutiny, or rapid inflation/costs.

### 5. Chart F (Dual-Axis Chart)
#### Lie Factor Evaluation: 
You cannot compute a meaningful Lie Factor for Chart F because the formula assumes a single, shared data scale. Dual axes superimpose two entirely different units and scales onto one canvas, making visual comparison mathematically invalid.  
#### What the data actually shows: 
Looking at case07_dual_axis_series.csv, revenue and cost follow independent trajectories that cannot be honestly correlated by overlapping two independent y-axes.
